# 01_Data_Cleaning

This notebook loads the daily log and formal exam results, checks data quality, and prepares clean datasets for further analysis.

# ----------Daily_Log----------

## 1. Loading data

```python
import pandas as pd
from pathlib import Path

sample_route = "../sample-data/sample_daily_log.csv"
authentic_route = "../authentic-data/daily_log.csv"

if Path(authentic_route).exists():
    df = pd.read_csv(authentic_route)    # DataFrame Formation
else:
    df = pd.read_csv(sample_route)

df["DATE"] = pd.to_datetime(df["DATE"])     # convert DATE column to datetime64[us] format
```

## 2. Basic checks

### i. Check missing values

```python
df.isna().sum()
```

```python
#print("Data missing values check done. Missing: someone to clean my heart. Wait, no, that's just missing you.")
```

### ii. Check duplicated values (DATE only)

```python
df["DATE"].duplicated().sum()
```

### iii. Check data structure

```python
df.info()
```

### iv. Check data scale (row, column (9))

```python
df.shape
```

## 3. Invalid values checks

### i. Define min_cols

```python
min_cols = [
    "SLEEP-MINS",
    "EXERCISE-MINS",
    "SCREEN-TIME-MINS",
    "MATH-MINS",
    "PHYSICS-MINS",
    "CHEMISTRY-MINS",
    "ENGLISH-MINS",
    "CHINESE-MINS"
]
```

### ii. Are there negative numbers? (I mean min_cols)

```python
(df[min_cols] < 0).sum()
```

### iii. Is there a day longer than 24 hours?

```python
(df[min_cols] > 1440).sum()
```

## 4. Import new column TOTAL-STUDY-MINS into DataFrame

```python
study_cols = [              # list of study-related columns
    "MATH-MINS",
    "PHYSICS-MINS",
    "CHEMISTRY-MINS",
    "ENGLISH-MINS",
    "CHINESE-MINS"
]

df["TOTAL-STUDY-MINS"] = df[study_cols].sum(axis=1)     # create a new column
```

## 5. Overall statistic

```python
df.describe(include="number")
```

# ----------Exam_Results----------

## 1. Loading data

```python
sample_route_2 = "../sample-data/sample_exam_results.csv"
authentic_route_2 = "../authentic-data/exam_results.csv"

if Path(authentic_route_2).exists():
    df_2 = pd.read_csv(authentic_route_2)
else:
    df_2 = pd.read_csv(sample_route_2)

df_2["DATE"] = pd.to_datetime(df_2["DATE"])
```

## 2. Basic checks

### i. Check missing values

```python
df_2.isna().sum()
```

### ii. Check duplicated values

```python
df_2.duplicated().sum()
```

### iii. Check data structure

```python
df_2.info()
```

### iv. Check data scale (row, column (10))

```python
df_2.shape
```

## 3. Invalid values checks

### i. Are there scores higher than the full scores?

```python
(df_2["SCORE"]>df_2["FULL-SCORE"]).sum()
```

### ii. Are there class ranks bigger than the class size?

```python
(df_2["CLASS-RANK"]>df_2["CLASS-SIZE"]).sum()
```

Or = 0?

```python
(df_2["CLASS-RANK"]==0).sum()
```

### iii. Are there grade ranks bigger than the grade size?

```python
(df_2["GRADE-RANK"]>df_2["GRADE-SIZE"]).sum()
```

Or = 0?

```python
(df_2["GRADE-RANK"]==0).sum()
```

### iv. Are there spelling mistakes?

##### Upper and strip each SUBJECT, TYPE and GRADE values

```python
df_2["SUBJECT"] = df_2["SUBJECT"].str.strip().str.upper()
df_2["TYPE"] = df_2["TYPE"].str.strip().str.upper()
df_2["GRADE"] = df_2["GRADE"].str.strip().str.upper()
```

##### Check

```python
print(df_2["SUBJECT"].value_counts(dropna=False))
print(df_2["TYPE"].value_counts(dropna=False))
print(df_2["GRADE"].value_counts(dropna=False))
```

The datasets were checked for missing values, duplicated dates, invalid values, and wrong data types.

No obvious data-quality issues were found in the current sample data.
